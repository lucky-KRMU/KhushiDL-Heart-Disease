# Data Preparation - Heart Disease Detection
---
## This Notebook is to prepare the final Data, which will be directly used in the Model Training

```python
# importing necessary modules
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.impute import KNNImputer
```

### Loading the data

```python
cleve_data = pd.read_csv('data/processed.cleveland.data', header=None)
swiss_data = pd.read_csv('data/processed.switzerland.data', header=None)
va_data = pd.read_csv('data/processed.va.data', header=None)
hungarian_data = pd.read_csv('data/processed.hungarian.data', header=None)
```

### Stacking / Merging the Data

```python
merged_data = pd.concat([cleve_data, swiss_data, va_data, hungarian_data], axis=0)
```

```python
merged_data.info()
```

```python
merged_data = merged_data.replace('?', pd.NA) # replacing ? with NaN
```

```python
merged_data.info()
```

```python

merged_data = merged_data.astype('Float64')
```

```python
merged_data.info()
```

## Splitting the Data

```python
# Splitting the data
merged_input = merged_data.iloc[: , [x for x in range(13)]].copy()
```

```python
merged_input.head()
```

```python
merged_data.head()
```

```python
merged_output = merged_data.iloc[:, [13]].copy()
```

```python
merged_output.head()
```

```python
merged_output.info()
```

### Test Train Split

```python
train_x, test_x, train_y, test_y = train_test_split(merged_input, merged_output, test_size=.3, random_state=28, shuffle=True)
```

```python
train_x.info()
```

```python
print(train_x.shape)
print(train_y.shape)
```

```python
print(test_x.shape)
print(test_y.shape)
```

```python
validation_x, test_x, validation_y, test_y = train_test_split(test_x, test_y, test_size=.5)
```

```python
print(validation_x.shape)
print(validation_y.shape)
print(test_x.shape)
print(test_y.shape)
```

## Imputation

```python
train_x.describe(include='all')
```

```python
# Imputation
imputer = KNNImputer(n_neighbors=5)

train_x = imputer.fit_transform(train_x) # fit to prevent data leakage

test_x = imputer.transform(test_x)
validation_x = imputer.transform(validation_x)
```

```python
# Converting data back to dataframe
train_x = pd.DataFrame(train_x, columns=None)
test_x = pd.DataFrame(test_x, columns=None)
validation_x = pd.DataFrame(validation_x, columns=None)
```

```python
train_x.head()
```

## Preparing Final Data

```python
train_x.info()
```

```python
# Stacking / Concatenating the Values
final_train_data = pd.concat([train_x, train_y], axis=0)
```

```python
final_train_data.shape
```

```python
final_train_data.info()
```

```python
final_validation_data = pd.concat([validation_x, validation_y], axis=0)
final_test_data = pd.concat([test_x, test_y], axis=0)
```

## Exporting the Data

```python
# Exporting the data 
final_train_data.to_csv('data/FinalDataSets/train.csv')
final_validation_data.to_csv('data/FinalDataSets/validation.csv')
final_test_data.to_csv('data/FinalDataSets/test.csv')
```
