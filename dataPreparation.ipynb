# Data Preparation - Heart Disease Detection

## This Notebook is to prepare the final Data, which will be directly used in the Model Training

```python
# importing necessary modules
import pandas as pd
```

### Loading the data

```python
cleve_data = pd.read_csv('data/processed.cleveland.data', header=None)
swiss_data = pd.read_csv('data/processed.switzerland.data', header=None)
va_data = pd.read_csv('data/processed.va.data', header=None)
hungarian_data = pd.read_csv('data/processed.hungarian.data', header=None)
```

### Stacking / Merging the Data

```python
merged_data = pd.concat([cleve_data, swiss_data, va_data, hungarian_data], axis=0)
```

```python
merged_data.info()
```

```python
merged_data = merged_data.replace('?', pd.NA) # replacing ? with NaN
```

```python
merged_data.info()
```

```python

merged_data = merged_data.astype('Float64')
```
